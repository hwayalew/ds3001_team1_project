```python

import requests
import pandas as pd
from google.colab import files

url = "https://www.fairfaxcounty.gov/euclid/rest/services/Health/PermittedFoodEstablishments/FeatureServer/1/query"

records = []
offset = 0

while True:
    params = {
        "where": "1=1",
        "outFields": "*",
        "returnGeometry": "false",
        "orderByFields": "OBJECTID ASC",
        "resultOffset": offset,
        "resultRecordCount": 2000,
        "f": "json"
    }

    response = requests.get(url, params=params)
    response.raise_for_status()
    data = response.json()

    if "error" in data:
        raise Exception(data["error"])

    batch = data["features"]
    if not batch:
        break

    records.extend(item["attributes"] for item in batch)
    offset += len(batch)

df = pd.DataFrame(records)

# Convert inspection dates to readable dates
df["INSPECTIONDATE"] = pd.to_datetime(
    df["INSPECTIONDATE"], unit="ms"
)

# Save and download CSV
df.to_csv("fairfax_food_inspections.csv", index=False)
print("Total records downloaded:", len(df))
files.download("fairfax_food_inspections.csv")
```
